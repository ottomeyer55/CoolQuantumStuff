```python
import numpy as np
import matplotlib.pyplot as plt

I_vals = np.array([1.5, 2.0, 2.5, 3.0, 3.5, 4.0])
P = np.array([270, 380, 620, 830, 1280, 1600])
p_error = 50

Isq = I_vals**2

plt.plot(Isq,P)
plt.errorbar(Isq, P, yerr=p_error, fmt='o', capsize=4, color='black')
plt.xlabel('I² (A²)')
plt.ylabel('Power, P (W)')
plt.title('P against I²')
plt.grid(alpha=0.3)
plt.show()
```
